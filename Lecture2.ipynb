# Groupby Operation

Often we want to compute summaries per category, for example: avarage flight time per airline or number of flights per day

For this we can use groupby(), that splits the data into groups.

groupby() only defines the groups – you must follow it with an aggregation such as
.mean(), .sum(), .count(), or .agg(…) to compute results.

```python
fligts.groupby("AIRLINE")['AIR_TIME'].mean().head(5)
```

Instead of selecting one column first, we can compute the mean of several numeric
columns per group and the result will be a DataFrame:

```python
fligts.groupby('AIRLINE')[['AIR_TIME', 'DISTANCE']].mean().head(5)
```

Note, that returned object is Series with unique "AIRLINE" values being in the index
column.

--------

### .size()

We can also compute the number of rows in each group using .size() directly after
groupby() (or .count() on any column):

```python
fligts.groupby('AIRLINE').size().head()
```

-------

### .agg()

Instead of one summary per column, .agg() lets us compute several stats at
once—per group.

```python
fligts.groupby('AIRLINE').agg( {
    'AIR_TIME': ['mean','std'],
    'DISTANCE': ['min'],
    'DESTINATION_AIRPORT': ['size', 'first', 'nunique']
    }
).head(5)
```

```python
fligts[ fligts['MONTH'].isin(range(6,9)) ].groupby('ORIGIN_AIRPORT').size().head(5)
```

# Extending Tables

## Creating a new column 

We can create new columns by performing operations on existing ones:

```python
fligts['DIST_MILES'] = fligts['DISTANCE'] * 0.621371
fligts.head(5)
```

We can also combine columns to compute new values. Here we compute speed by
dividing distance by air time:

```python
fligts['SPEED_MPH'] = (
    fligts['DIST_MILES'] / fligts['AIR_TIME'] * 60
)

fligts.head(5)
```

------

## Drop a Column

To delete column, you can use **.drop()** function and provide column names as a list:


This function return a new DataFrame by default. Use **inplace=True** to modify the
DataFrame directly.

```python
fligts.drop(columns = ['DIST_MILES', 'SPEED_MPH'], inplace= True)
fligts.head(5)
```
