## np.unique() 
### with parameter return_counts

```python
x = np.array(['DC', 'Alabama', 'Florida', 'Florida', 'DC', 'DC'])
unique_values = np.unique(x, return_counts=True)

print('Values: ', unique_values[0])
print('Counts: ', unique_values[1])
```

## isna

The *isna* function, applied to the column of the dataset, returns a logical vector that tells us which
entries are nan. Assign this logical vector to an object called ind and determine how many nans does
age have.

```python
ind = titanic['age'].isna()
ind.sum()
```

```python
ind_np = np.isnan(titanic['age'].values)

(
titanic['age'].values[~ind_np].mean(), 
titanic['age'].mean()
)
```
