8. You are given the starwars data table which contains data of different characters appearing in the
Star Wars movies. A few rows from the data table are shown below.

```python
star_wars_df = pd.read_csv('../Datasets/Exercise2_datasets/star_wars_character_dataset.csv',
                    usecols = ["name", "height", "mass", "birth_year", "sex", "homeworld", "species"])

star_wars_df = star_wars_df.rename(columns = {'birth_year': 'age'})
star_wars_df.head()
```

a) Find the number of individuals from each species with a height over 200.

```python
star_wars_df[star_wars_df['height'] > 200].groupby('species').size()
```

b) Find the mean age of each species and filter the results for species that have mean age greater than
70

```python
mean_age = star_wars_df.groupby('species')['age'].mean()

mean_age[ mean_age > 70]
```

c) Which are the top-10 species with the highest average body-mass-index (BMI)? BMI is given by
mass $\text{kg}/\text{height}^2$
(meters). Hint: leaving 0’s in the denominator or NANs when calculating the
mean could lead to errors! 

Also make sure height and mass are in the correct unit.

```python
star_wars_df2 = star_wars_df[ ~star_wars_df['height'].isin([0, pd.NA])]

star_wars_df2['height'] = star_wars_df2['height'] / 100
star_wars_df2['BMI'] = star_wars_df2['mass'] / star_wars_df2['height']**2

star_wars_df2.groupby('species')['BMI'].mean().sort_values(ascending= False).head(10).index
```

1. Add a new column called High_Rating to the DataFrame ratings_df. The column has an integer
1 for all observations with a rating value higher than 7. Hint: You can create a boolean variable and
then convert it to int.

```python
ratings_dt['High_Rating'] = (ratings_dt['Book_Rating'] > 7).astype('Int64')

ratings_dt.head()
```

2. How many observations are considered to be a high ranking? What is the proportion of high ranked
observations among all observations?

```python
count = ratings_dt['High_Rating'].sum()
proportion = count / (ratings_dt.size - count)
print('High Ratings: {}\nProportion: {:.2f}%'.format(count, proportion * 100)) 
```

6. What is the title of the first published book with the highest ranking? Hint: assume that year of
publication being zero is an error.

```python
ratings_dt[(~ratings_dt['Year_Of_Publication'].isin([np.nan, 0])) & (ratings_dt['Book_Rating'] == 10)] \
.sort_values(by = 'Year_Of_Publication', ascending= True).iloc[0]['Book_Title']
```

7. In which year was the book with the largest number of ratings published? For simplicity we consider
